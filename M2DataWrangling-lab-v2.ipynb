### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Non-Null Count": df.notnull().sum(),
    "Missing Values": df.isnull().sum()
})

summary
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
df.describe()
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# View unique country values
df["Country"].unique()
# Count occurrences of each country
df["Country"].value_counts(dropna=False)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
df["Country"].value_counts(dropna=False)
country_mapping = {
    "United States of America": "United States",
    "USA": "United States",
    "US": "United States",
    "UK": "United Kingdom",
    "England": "United Kingdom",
    "Scotland": "United Kingdom",
    "Russia": "Russian Federation",
    "Viet Nam": "Vietnam"
}

df["Country"] = df["Country"].replace(country_mapping)
df["Country"] = df["Country"].replace({"Prefer not to say": pd.NA})
df["EdLevel"].value_counts(dropna=False)
edlevel_mapping = {
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's degree",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master's degree",
    "Doctoral degree (Ph.D., Ed.D., etc.)": "Doctoral degree",
    "Some college/university study without earning a degree": "Some college",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "Secondary school",
    "Primary/elementary school": "Primary school"
}

df["EdLevel"] = df["EdLevel"].replace(edlevel_mapping)
df["Country"].value_counts().head(10)
df["EdLevel"].value_counts()
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
# One-hot encode the Employment column
employment_dummies = pd.get_dummies(df["Employment"], prefix="Employment")

# Append the encoded columns to the original DataFrame
df = pd.concat([df, employment_dummies], axis=1)
df.head()
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
# Identify columns with the highest number of missing values
missing_counts = df.isnull().sum().sort_values(ascending=False)

missing_counts.head()
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
# Impute numerical column with median
median_val = df["ConvertedCompYearly"].median()
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(median_val)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
# Identify the most frequent value
most_frequent_remote = df["RemoteWork"].mode()[0]

# Impute missing values
df["RemoteWork"] = df["RemoteWork"].fillna(most_frequent_remote)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
# Apply Min-Max Scaling to normalize ConvertedCompYearly
min_val = df["ConvertedCompYearly"].min()
max_val = df["ConvertedCompYearly"].max()

df["ConvertedCompYearly_MinMax"] = (
    (df["ConvertedCompYearly"] - min_val) / (max_val - min_val)
)

# Verify the normalization
df["ConvertedCompYearly_MinMax"].describe()
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
import numpy as np

# Log-transform ConvertedCompYearly to reduce skewness
# Use log1p to safely handle zero values
df["ConvertedCompYearly_Log"] = np.log1p(df["ConvertedCompYearly"])
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
df["YearsCodePro"].unique()
# Convert YearsCodePro to numeric
df["YearsCodePro_Num"] = df["YearsCodePro"].replace({
    "Less than 1 year": 0,
    "More than 50 years": 50
})

df["YearsCodePro_Num"] = pd.to_numeric(df["YearsCodePro_Num"], errors="coerce")
```
